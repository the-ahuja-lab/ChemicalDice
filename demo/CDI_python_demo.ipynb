### Input Prepration

```python
import pandas as pd
import os


data = {
    'SMILES': [
        'CCO',                          # Ethanol
        'c1ccccc1',                     # Benzene
        'CC(=O)Oc1ccccc1C(=O)O',        # Aspirin
        'CN1C=NC2=C1C(=O)N(C(=O)N2C)C', # Caffeine
        'C(C1C(C(C(C(O1)O)O)O)O)O'      # Glucose
    ]
}


df = pd.DataFrame(data)


filename = "smiles.csv"
df.to_csv(filename, index=False)
```
